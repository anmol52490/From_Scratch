```python
import time
from threading import Thread

COUNT = 50000000

def countdown(n):
    while n>0:
        n -= 1

start = time.time()
countdown(COUNT)
end = time.time()

print('Time taken in seconds -', end - start)
```

```python
import time
from threading import Thread

COUNT = 50000000

def countdown(n):
    while n>0:
        n -= 1

t1 = Thread(target=countdown, args=(COUNT//2,))
t2 = Thread(target=countdown, args=(COUNT//2,))

start = time.time()
t1.start()
t2.start()
t1.join()
t2.join()
end = time.time()

print('Time taken in seconds -', end - start)
```

```python
import os
import threading
os.cpu_count()

def worker():
    i = 0
    while True:
        i+=1

def main():
    n = os.cpu_count() or 1
    print(f"spawning {n} threads")

    threads = []

    for _ in range(n):
        t = threading.Thread(target = worker)
        t.start()
        threads.append(t)

    for t in threads:
        t.join()

if __name__ == "__main__":
    main()
```

```python
#MultiThreading

import threading 
import time

def print_message(name, count):
    for i in range(count):
        print(f"{name} says : {i}")
        time.sleep(0.5)

def main():
    threads = []
    for i in range(3):
        thread = threading.Thread(target=print_message, args = (f"thread - {i+1}", 3))
        thread.start()
        threads.append(thread)
    thread.join()
    print("\nMain thread finished.")
    return threads
    
if __name__ == '__main__': 
    t = main()
    print(t)
```

```python
#MultiThreading

import threading 
import time
lock = threading.Lock()
def print_message(name, count): 
    for i in range(count):
        with lock:
            print(f"{name} says : {i}")
            time.sleep(1)

def main():
    threads = []
    for i in range(3):
        thread = threading.Thread(target=print_message, args = (f"thread - {i+1}", 3))
        threads.append(thread)
        thread.start()
        
    for thread in threads:
        thread.join()
    print("\nMain thread finished.")
    return threads
    
if __name__ == '__main__':
    t = main()
    print(t)
```

```python
# daemon thread = stops when main thread stops

import threading 
import time

def infinite_task():
    while True:
        print("Running")
        time.sleep(1)

def main():
    daemon_thread = threading.Thread(target = infinite_task)
    daemon_thread.daemon = True
    daemon_thread.start()
    time.sleep(3)
    print(" Main thread finished")


if __name__ == '__main__':
    main()
```

```python
# Thread with sync using lock but GIL kinda handels all this
import threading 

counter = 0
counter_lock = threading.Lock()

def increment():
    global counter
    with counter_lock:
        for _ in range(100000):
            counter+=1

def main():
    global counter

    threads = []
    for i in range(3):
        thread = threading.Thread(target = increment)
        threads.append(thread)
        thread.start()
```

```python
# Python thread queue ( when trying to pass data between threads)
import queue
import threading
import time

def producer(queue):
    for i in range(5):
        item = f"item- {i}"
        print(f"Producing {item}")
        queue.put(item)
        time.sleep(1)
    queue.put(None)


def consumer(queue):
    while True:
        item = queue.get()
        if item is None:
            break
        print(f"consuming {item}")
        time.sleep(2)
def main():
    q = queue.Queue()
    producer_thread = threading.Thread(target=producer, args= (q,))
    consumer_thread = threading.Thread(target=consumer, args=( q,))
    producer_thread.start()
    consumer_thread.start()
    producer_thread.join()
    consumer_thread.join()

if __name__ == "__main__":
    main()
```
